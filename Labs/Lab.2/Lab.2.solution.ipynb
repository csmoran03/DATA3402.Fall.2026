````python
# Write your solution here

size = 3
empty = 0

row_names = list("ABCDEFGHIJKLMNOPQRSTUVWXYZ")[:size]
row_map = dict(zip(row_names, range(size)))

column_names = list(map(str, range(1, size + 1)))
column_map = dict(zip(column_names, range(size)))

def draw_board(board):
    print("\n ", end=" ")
    for j in range(size):
        print(column_names[j], end=" ")
    print()

    for i in range(size):
        print(row_names[i], end=" ")
        for j in range(size):
            if board[i][j] == 1:
                mark = 'X'
            elif board[i][j] == 2:
                mark = 'O'
            else:
                mark = ' '
            print(f"[{mark}]", end=" ")
        print()
    print()

def move_piece(board, player, location, verbose=True):
    x, y = location
    move_possible = 0 <= x < size and 0 <= y < size

    if not move_possible:
        if verbose: print("Move is off of board.")
        return False

    if board[x][y] == empty:
        board[x][y] = player
        if verbose: print("Moved.")
        return True
    else:
        if verbose: print("Space is already taken. Move not possible.")
        return False

def make_string_move(board, player, location_string):
    row_char= location_string[0].upper()
    col_char =location_string[1:]

    x= row_map[row_char]
    y= column_map[col_char]

    return move_piece(board, player, (x, y))

def player_turn(board, player):
    valid_move= False

    while not valid_move:
        mark = 'X' if player == 1 else 'O'
        location_string = input(f"Player {player} ({mark}), enter your move (e.g., 'A1'): ")

        try:
            valid_move = make_string_move(board, player, location_string)
        except (KeyError, IndexError):
            print("Invalid input format. Please enter a valid row and column (e.g., 'A1').")


def check_winner(board, player):
    for row in board:
        if all(cell == player for cell in row):
            return True

    # Check columns
    for col in range(size):
        if all(board[row][col] == player for row in range(size)):
            return True

    # Check diagonals
    if all(board[i][i] == player for i in range(size)):
        return True
    if all(board[i][size - 1 - i] == player for i in range(size)):
        return True

    return False

def check_draw(board):
    for row in board:
        if empty in row:
            return False
    return True

def evaluate_board(board, computer, human):
    if check_winner(board, computer):
        return 10
    elif check_winner(board, human):
        return -10
    else:
        return 0

def minimax(board, depth, is_maximizing, computer, human):

    score = evaluate_board(board, computer, human)

    if score == 10:
        return score + depth
    if score == -10:
        return score - depth
    if check_draw(board):
        return 0
    if depth == 0:
        return 0

    if is_maximizing:
        best_score = -float('inf')
        for r in range(size):
            for c in range(size):
                if board[r][c] == empty:
                    board[r][c] = computer
                    current_score = minimax(board, depth - 1, False, computer, human)
                    board[r][c] = empty
                    best_score = max(score, best_score, current_score)
        return best_score

    else:
        best_score = float('inf')
        for r in range(size):
            for c in range(size):
                if board[r][c] == empty:
                    board[r][c] = human
                    current_score = minimax(board, depth - 1, True, computer, human)
                    board[r][c] = empty
                    best_score = min(score, best_score, current_score)
        return best_score

def get_best_move(board, computer_player, max_depth):
    human_player = 1 if computer_player == 2 else 2
    best_score = -float('inf')
    best_move = None

    for r in range(size):
        for c in range(size):
            if board[r][c] == empty:
                board[r][c] = computer_player
                move_score = minimax(board, max_depth - 1, False, computer_player, human_player)
                board[r][c] = empty

                if move_score > best_score:
                    best_score = move_score
                    best_move = (r, c)

    return best_move

def computer_turn(board, computer_player, max_depth=9):
    print("Computer is thinking (simulating futures)...")
    time.sleep(0.5)

    best_move= get_best_move(board, computer_player, max_depth)

    if best_move:
        r, c = best_move
        board[r][c] = computer_player
        print(f"Computer plays at {row_names[r]}{column_names[c]}!")

def play_game_vs_minimax():
    board= [[empty for _ in range(size)] for _ in range(size)]
    current_player = 1

    print("Welcome to Tic-Tac-Toe vs Minimax AI!")
    print("1. You go first (Player 1 - X)")
    print("2. Computer goes first (Player 1 - X)")
    choice = input("Enter 1 or 2: ")

    human = 1 if choice == '1' else 2
    computer = 2 if human == 1 else 1

    max_depth = 9

    while True:
        draw_board(board)

        if current_player == human:
            player_turn(board, human)
        else:
            computer_turn(board, computer, max_depth)

        if check_winner(board, current_player):
            draw_board(board)
            if current_player == human:
                print(" You win! (Wait, how did you beat Minimax?!)")
            else:
                print("The Computer wins! (Exhaustive search reigns supreme!)")
            break

        if check_draw(board):
            draw_board(board)
            print("It's a draw! Perfect play from both sides.")
            break

        current_player = 2 if current_player == 1 else 1

````

*Exercise 13:* Make the 2 computer players play each-other for 10 games on a 3x3, then 4x4, then 5x5 grid. Set the max depth so that the games only take seconds. Measure the "smarter" player's win rate for each grid.

````python
# Write your solution here
import time

def check_winner(board, player, size):
    for row in board:
        if all(cell == player for cell in row): return True
    for col in range(size):
        if all(board[row][col] == player for row in range(size)): return True
    if all(board[i][i] == player for i in range(size)): return True
    if all(board[i][size - 1 - i] == player for i in range(size)): return True
    return False

def check_draw(board):
    for row in board:
        if 0 in row: return False
    return True

def basic_ai_move(board, computer_player, size):
    opponent = 1 if computer_player == 2 else 2

    # 1. Win
    for r in range(size):
        for c in range(size):
            if board[r][c] == 0:
                board[r][c]= computer_player
                if check_winner(board, computer_player, size):
                    board[r][c] = 0
                    return (r, c)
                board[r][c] = 0

    # 2. Block
    for r in range(size):
        for c in range(size):
            if board[r][c] == 0:
                board[r][c] =opponent
                if check_winner(board, opponent, size):
                    board[r][c] = 0
                    return (r, c)
                board[r][c] = 0

    # 3. Center
    if board[size//2][size//2] == 0:
        return (size//2, size//2)

    empty_spots = [(r, c) for r in range(size) for c in range(size) if board[r][c] == 0]
    return random.choice(empty_spots) if empty_spots else None


def evaluate_board(board, computer, human, size):
    if check_winner(board, computer, size): return 10
    if check_winner(board, human, size): return -10
    return 0

def minimax(board, depth, is_maximizing, computer, human, size):
    score = evaluate_board(board, computer, human, size)

    if score== 10: return score + depth
    if score == -10: return score - depth
    if check_draw(board) or depth == 0: return 0

    if is_maximizing:
        best_score = -float('inf')
        for r in range(size):
            for c in range(size):
                if board[r][c] == 0:
                    board[r][c] = computer
                    best_score = max(best_score, minimax(board, depth - 1, False, computer, human, size))
                    board[r][c] = 0
        return best_score
    else:
        best_score = float('inf')
        for r in range(size):
            for c in range(size):
                if board[r][c] == 0:
                    board[r][c] = human
                    best_score = min(best_score, minimax(board, depth - 1, True, computer, human, size))
                    board[r][c] = 0
        return best_score

def minimax_move(board, computer_player, max_depth, size):
    human_player = 1 if computer_player == 2 else 2
    best_score = -float('inf')
    best_moves = []

    for r in range(size):
        for c in range(size):
            if board[r][c] == 0:
                board[r][c] = computer_player
                move_score = minimax(board, max_depth - 1, False, computer_player, human_player, size)
                board[r][c] = 0

                if move_score > best_score:
                    best_score = move_score
                    best_moves = [(r, c)]
                elif move_score == best_score:
                    best_moves.append((r, c))

    return random.choice(best_moves) if best_moves else None


def simulate():
    grid_sizes = [3, 4, 5]
    games_per_size = 10

    depth_limits = {
        3: 9,
        4: 3,
        5: 2
    }

    for size in grid_sizes:
        print(f"--- Simulating 10 games on {size}x{size} Grid ---")
        max_d = depth_limits[size]
        print(f"Minimax Depth limit set to: {max_d}")

        minimax_wins = 0
        basic_wins = 0
        draws = 0

        start_time = time.time()

        for game in range(games_per_size):
            board = [[0 for _ in range(size)] for _ in range(size)]

            basic_player = 1 if game % 2 == 0 else 2
            minimax_player = 2 if basic_player == 1 else 1

            current_turn = 1

            while True:
                if current_turn == basic_player:
                    move = basic_ai_move(board, basic_player, size)
                else:
                    move = minimax_move(board, minimax_player, max_d, size)

                board[move[0]][move[1]] = current_turn

                if check_winner(board, current_turn, size):
                    if current_turn == minimax_player:
                        minimax_wins += 1
                    else:
                        basic_wins += 1
                    break

                if check_draw(board):
                    draws += 1
                    break

                current_turn= 2 if current_turn == 1 else 1

        elapsed = round(time.time() - start_time, 2)
        print(f"Time taken: {elapsed} seconds")
        print(f"Smarter AI (Minimax) Win Rate: {(minimax_wins/games_per_size)*100}%")
        print(f"Basic AI Wins: {basic_wins}")
        print(f"Draws: {draws}\n")

````
