```python
import math
class Counter:
    def __init__(self, max_value: int):
        if max_value < 0:
            raise ValueError("Maximum value must be non-negative.")
        self.max_value = max_value
        self.value = 0

    def increment(self) -> None:
        if self.value < self.max_value:
            self.value += 1
        else:
            print(f"Error: Cannot increment beyond the maximum value of {self.max_value}.")

    def reset(self) -> None:
        self.value = 0

    def __repr__(self) -> str:
        return f"Counter(value={self.value}, max_value={self.max_value})"


if __name__ == "__main__":
    counter = Counter(max_value=2)
    print(counter)

    counter.increment()
    counter.increment()
    print(counter)
    counter.increment()
    print(counter)
    counter.reset()
    print(counter)
```

```python
class Rectangle:
    def __init__(self, length: float, width: float, x: float = 0.0, y: float = 0.0):
        if length < 0 or width < 0:
            raise ValueError("Length and width must be non-negative.")
        self.__length = length
        self.__width = width
        self.__x = x
        self.__y = y

    def get_length(self) -> float:
        return self.__length

    def get_width(self) -> float:
        return self.__width

    def get_x(self) -> float:
        return self.__x

    def get_y(self) -> float:
        return self.__y

    def area(self) -> float:
        return self.__length * self.__width

    def perimeter(self) -> float:
        return 2 * (self.__length + self.__width)

    def __repr__(self) -> str:
        return (
            f"Rectangle(length={self.__length}, width={self.__width}, "
            f"x={self.__x}, y={self.__y})"
        )


if __name__ == "__main__":
    rect = Rectangle(length=10.0, width=5.0, x=2.0, y=3.0)
    print(rect)

    print(f"Length: {rect.get_length()}, Width: {rect.get_width()}")
    print(f"Corner coordinates: ({rect.get_x()}, {rect.get_y()})")
    print(f"Area: {rect.area()}")
    print(f"Perimeter: {rect.perimeter()}")
```

9. Add a function in the base class of the object classes that returns true/false testing that the object overlaps with another object.

```python

class Shape:

    def get_x(self) -> float:
        raise NotImplementedError("Subclasses must implement get_x().")

    def get_y(self) -> float:
        raise NotImplementedError("Subclasses must implement get_y().")

    def area(self) -> float:
        raise NotImplementedError("Subclasses must implement area().")

    def perimeter(self) -> float:
        raise NotImplementedError("Subclasses must implement perimeter().")

    def get_perimeter_points(self) -> list[tuple[float, float]]:
        raise NotImplementedError("Subclasses must implement get_perimeter_points().")

    def is_inside(self, px: float, py: float) -> bool:
        raise NotImplementedError("Subclasses must implement is_inside().")

    def overlaps(self, other: "Shape") -> bool:

        for px, py in self.get_perimeter_points():
            if other.is_inside(px, py):
                return True


        for px, py in other.get_perimeter_points():
            if self.is_inside(px, py):
                return True

        return False


class Rectangle(Shape):

    def __init__(self, length: float, width: float, x: float = 0.0, y: float = 0.0):
        if length < 0 or width < 0:
            raise ValueError("Length and width must be non-negative.")
        self.__length = length
        self.__width = width
        self.__x = x
        self.__y = y

    def get_length(self) -> float:
        return self.__length

    def get_width(self) -> float:
        return self.__width

    def get_x(self) -> float:
        return self.__x

    def get_y(self) -> float:
        return self.__y

    def area(self) -> float:
        return self.__length * self.__width

    def perimeter(self) -> float:
        return 2 * (self.__length + self.__width)

    def get_perimeter_points(self) -> list[tuple[float, float]]:
        x, y = self.__x, self.__y
        L, W = self.__length, self.__width
        corners = [
            (x, y),
            (x + L, y),
            (x + L, y + W),
            (x, y + W), ]

        points: list[tuple[float, float]] = []
        for i in range(4):
            x1, y1 = corners[i]
            x2, y2 = corners[(i + 1) % 4]
            for step in range(4):
                t = step / 4.0
                points.append((x1 + t * (x2 - x1), y1 + t * (y2 - y1)))
        return points

    def is_inside(self, px: float, py: float) -> bool:
        return (
            self.__x <= px <= self.__x + self.__length
            and self.__y <= py <= self.__y + self.__width
        )

    def __repr__(self) -> str:
        return (
            f"Rectangle(length={self.__length}, width={self.__width}, "
            f"x={self.__x}, y={self.__y})"
        )


class Circle(Shape):

    def __init__(self, radius: float, x: float = 0.0, y: float = 0.0):
        if radius < 0:
            raise ValueError("Radius must be non-negative.")
        self.__radius = radius
        self.__x = x
        self.__y = y

    def get_radius(self) -> float:
        return self.__radius

    def get_x(self) -> float:
        return self.__x

    def get_y(self) -> float:
        return self.__y

    def area(self) -> float:
        return math.pi * (self.__radius ** 2)

    def perimeter(self) -> float:
        return 2 * math.pi * self.__radius

    def get_perimeter_points(self) -> list[tuple[float, float]]:
        points: list[tuple[float, float]] = []
        num_points = 16
        for i in range(num_points):
            angle = (2 * math.pi * i) / num_points
            px = self.__x + self.__radius * math.cos(angle)
            py = self.__y + self.__radius * math.sin(angle)
            points.append((px, py))
        return points

    def is_inside(self, px: float, py: float) -> bool:
        dist_squared = (px - self.__x) ** 2 + (py - self.__y) ** 2
        return dist_squared <= self.__radius ** 2

    def __repr__(self) -> str:
        return f"Circle(radius={self.__radius}, x={self.__x}, y={self.__y})"


class Triangle(Shape):

    def __init__(
        self,
        side_a: float,
        side_b: float,
        side_c: float,
        x: float = 0.0,
        y: float = 0.0,
    ):
        if side_a <= 0 or side_b <= 0 or side_c <= 0:
            raise ValueError("All side lengths must be positive.")
        if (
            side_a + side_b <= side_c
            or side_a + side_c <= side_b
            or side_b + side_c <= side_a
        ):
            raise ValueError(
                "Invalid side lengths: violates the triangle inequality theorem."
            )

        self.__side_a = side_a
        self.__side_b = side_b
        self.__side_c = side_c
        self.__x = x
        self.__y = y

    def get_side_a(self) -> float:
        return self.__side_a

    def get_side_b(self) -> float:
        return self.__side_b

    def get_side_c(self) -> float:
        return self.__side_c

    def get_x(self) -> float:
        return self.__x

    def get_y(self) -> float:
        return self.__y

    def __get_vertices(self) -> list[tuple[float, float]]:
        a, b, c = self.__side_a, self.__side_b, self.__side_c
        v1 = (self.__x, self.__y)
        v2 = (self.__x + a, self.__y)

        cos_a = (a**2 + b**2 - c**2) / (2 * a * b)
        cos_a = max(-1.0, min(1.0, cos_a))
        sin_a = math.sqrt(1.0 - cos_a**2)
        v3 = (self.__x + b * cos_a, self.__y + b * sin_a)
        return [v1, v2, v3]

    def area(self) -> float:
        s = self.perimeter() / 2.0
        return math.sqrt(
            s * (s - self.__side_a) * (s - self.__side_b) * (s - self.__side_c)
        )

    def perimeter(self) -> float:
        return self.__side_a + self.__side_b + self.__side_c

    def get_perimeter_points(self) -> list[tuple[float, float]]:
        vertices = self.__get_vertices()
        points: list[tuple[float, float]] = []
        for i in range(3):
            x1, y1 = vertices[i]
            x2, y2 = vertices[(i + 1) % 3]
            for step in range(5):
                t = step / 5.0
                points.append((x1 + t * (x2 - x1), y1 + t * (y2 - y1)))
        return points

    def is_inside(self, px: float, py: float) -> bool:
        (x1, y1), (x2, y2), (x3, y3) = self.__get_vertices()

        def cross_product(ax: float, ay: float, bx: float, by: float) -> float:
            return (bx - ax) * (py - ay) - (by - ay) * (px - ax)

        d1 = cross_product(x1, y1, x2, y2)
        d2 = cross_product(x2, y2, x3, y3)
        d3 = cross_product(x3, y3, x1, y1)

        eps = 1e-9
        has_neg = (d1 < -eps) or (d2 < -eps) or (d3 < -eps)
        has_pos = (d1 > eps) or (d2 > eps) or (d3 > eps)
        return not (has_neg and has_pos)

    def __repr__(self) -> str:
        return (
            f"Triangle(side_a={self.__side_a}, side_b={self.__side_b}, "
            f"side_c={self.__side_c}, x={self.__x}, y={self.__y})"
        )


if __name__ == "__main__":
    rect = Rectangle(length=6.0, width=4.0, x=0.0, y=0.0)
    circle_near = Circle(radius=3.0, x=5.0, y=2.0)
    circle_far = Circle(radius=2.0, x=20.0, y=20.0)
    tri = Triangle(side_a=3.0, side_b=4.0, side_c=5.0, x=1.0, y=1.0)

    print(f"rect overlaps circle_near? {rect.overlaps(circle_near)}")
    print(f"rect overlaps circle_far?  {rect.overlaps(circle_far)}")
    print(f"rect overlaps tri?         {rect.overlaps(tri)}")
    print(f"tri overlaps circle_far?   {tri.overlaps(circle_far)}")
```

```python

class Canvas:
    def __init__(self, width, height):
        self.width = width
        self.height = height
        self.data = [[' '] * width for i in range(height)]

    def set_pixel(self, row, col, char='*'):
        self.data[row][col] = char

    def get_pixel(self, row, col):
        return self.data[row][col]

    def clear_canvas(self):
        self.data = [[' '] * self.width for i in range(self.height)]

    def v_line(self, x, y, h, **kargs):
        for i in range(x, x + h):
            self.set_pixel(i, y, **kargs)

    def h_line(self, x, y, w, **kargs):
        for i in range(y, y + w):
            self.set_pixel(x, i, **kargs)

    def line(self, x1, y1, x2, y2, **kargs):
        slope = (x2 - x1) / (y2 - y1)
        for y in range(y1, y2):
            x = x1 + int(slope * (y - y1))
            self.set_pixel(x, y, **kargs)

    def display(self):
        print("\n".join(["".join(row) for row in self.data]))


class Shape:

    def get_x(self) -> float:
        raise NotImplementedError("Subclasses must implement get_x().")

    def get_y(self) -> float:
        raise NotImplementedError("Subclasses must implement get_y().")

    def area(self) -> float:
        raise NotImplementedError("Subclasses must implement area().")

    def perimeter(self) -> float:
        raise NotImplementedError("Subclasses must implement perimeter().")

    def get_perimeter_points(self) -> list[tuple[float, float]]:
        raise NotImplementedError("Subclasses must implement get_perimeter_points().")

    def is_inside(self, px: float, py: float) -> bool:
        raise NotImplementedError("Subclasses must implement is_inside().")

    def overlaps(self, other: "Shape") -> bool:
        for px, py in self.get_perimeter_points():
            if other.is_inside(px, py):
                return True
        for px, py in other.get_perimeter_points():
            if self.is_inside(px, py):
                return True
        return False

    def paint(self, canvas: Canvas, **kargs) -> None:
        raise NotImplementedError("Subclasses must implement paint().")


class Rectangle(Shape):

    def __init__(self, length: float, width: float, x: float = 0.0, y: float = 0.0):
        if length < 0 or width < 0:
            raise ValueError("Length and width must be non-negative.")
        self.__length = length
        self.__width = width
        self.__x = x
        self.__y = y

    def get_length(self) -> float:
        return self.__length

    def get_width(self) -> float:
        return self.__width

    def get_x(self) -> float:
        return self.__x

    def get_y(self) -> float:
        return self.__y

    def area(self) -> float:
        return self.__length * self.__width

    def perimeter(self) -> float:
        return 2 * (self.__length + self.__width)

    def get_perimeter_points(self) -> list[tuple[float, float]]:
        x, y = self.__x, self.__y
        L, W = self.__length, self.__width
        corners = [
            (x, y),
            (x + L, y),
            (x + L, y + W),
            (x, y + W),
        ]
        points: list[tuple[float, float]] = []
        for i in range(4):
            x1, y1 = corners[i]
            x2, y2 = corners[(i + 1) % 4]
            for step in range(4):
                t = step / 4.0
                points.append((x1 + t * (x2 - x1), y1 + t * (y2 - y1)))
        return points

    def is_inside(self, px: float, py: float) -> bool:
        return (
            self.__x <= px <= self.__x + self.__length
            and self.__y <= py <= self.__y + self.__width
        )

    def paint(self, canvas: Canvas, **kargs) -> None:
        r = int(round(self.__x))
        c = int(round(self.__y))
        h = int(round(self.__length))
        w = int(round(self.__width))

        canvas.v_line(r, c, h, **kargs)
        canvas.v_line(r, c + w, h, **kargs)
        canvas.h_line(r, c, w, **kargs)
        canvas.h_line(r + h, c, w + 1, **kargs)
    def __repr__(self) -> str:
        return (
            f"Rectangle(length={self.__length}, width={self.__width}, "
            f"x={self.__x}, y={self.__y})"
        )


class Circle(Shape):

    def __init__(self, radius: float, x: float = 0.0, y: float = 0.0):
        if radius < 0:
            raise ValueError("Radius must be non-negative.")
        self.__radius = radius
        self.__x = x
        self.__y = y

    def get_radius(self) -> float:
        return self.__radius

    def get_x(self) -> float:
        return self.__x

    def get_y(self) -> float:
        return self.__y

    def area(self) -> float:
        return math.pi * (self.__radius ** 2)

    def perimeter(self) -> float:
        return 2 * math.pi * self.__radius

    def get_perimeter_points(self) -> list[tuple[float, float]]:
        points: list[tuple[float, float]] = []
        num_points = 16
        for i in range(num_points):
            angle = (2 * math.pi * i) / num_points
            px = self.__x + self.__radius * math.cos(angle)
            py = self.__y + self.__radius * math.sin(angle)
            points.append((px, py))
        return points

    def is_inside(self, px: float, py: float) -> bool:
        return (px - self.__x) ** 2 + (py - self.__y) ** 2 <= self.__radius ** 2

    def paint(self, canvas: Canvas, **kargs) -> None:
        for px, py in self.get_perimeter_points():
            r = int(round(px))
            c = int(round(py))
            if 0 <= r < canvas.height and 0 <= c < canvas.width:
                canvas.set_pixel(r, c, **kargs)

    def __repr__(self) -> str:
        return f"Circle(radius={self.__radius}, x={self.__x}, y={self.__y})"


class Triangle(Shape):

    def __init__(
        self,
        side_a: float,
        side_b: float,
        side_c: float,
        x: float = 0.0,
        y: float = 0.0,
    ):
        if side_a <= 0 or side_b <= 0 or side_c <= 0:
            raise ValueError("All side lengths must be positive.")
        if (
            side_a + side_b <= side_c
            or side_a + side_c <= side_b
            or side_b + side_c <= side_a
        ):
            raise ValueError(
                "Invalid side lengths: violates the triangle inequality theorem."
            )

        self.__side_a = side_a
        self.__side_b = side_b
        self.__side_c = side_c
        self.__x = x
        self.__y = y

    def get_side_a(self) -> float:
        return self.__side_a

    def get_side_b(self) -> float:
        return self.__side_b

    def get_side_c(self) -> float:
        return self.__side_c

    def get_x(self) -> float:
        return self.__x

    def get_y(self) -> float:
        return self.__y

    def __get_vertices(self) -> list[tuple[float, float]]:
        a, b, c = self.__side_a, self.__side_b, self.__side_c
        v1 = (self.__x, self.__y)
        v2 = (self.__x + a, self.__y)

        cos_a = (a**2 + b**2 - c**2) / (2 * a * b)
        cos_a = max(-1.0, min(1.0, cos_a))
        sin_a = math.sqrt(1.0 - cos_a**2)
        v3 = (self.__x + b * cos_a, self.__y + b * sin_a)
        return [v1, v2, v3]

    def area(self) -> float:
        s = self.perimeter() / 2.0
        return math.sqrt(
            s * (s - self.__side_a) * (s - self.__side_b) * (s - self.__side_c)
        )

    def perimeter(self) -> float:
        return self.__side_a + self.__side_b + self.__side_c

    def get_perimeter_points(self) -> list[tuple[float, float]]:
        vertices = self.__get_vertices()
        points: list[tuple[float, float]] = []
        for i in range(3):
            x1, y1 = vertices[i]
            x2, y2 = vertices[(i + 1) % 3]
            for step in range(5):
                t = step / 5.0
                points.append((x1 + t * (x2 - x1), y1 + t * (y2 - y1)))
        return points

    def is_inside(self, px: float, py: float) -> bool:
        (x1, y1), (x2, y2), (x3, y3) = self.__get_vertices()

        def cross_product(ax: float, ay: float, bx: float, by: float) -> float:
            return (bx - ax) * (py - ay) - (by - ay) * (px - ax)

        d1 = cross_product(x1, y1, x2, y2)
        d2 = cross_product(x2, y2, x3, y3)
        d3 = cross_product(x3, y3, x1, y1)

        eps = 1e-9
        has_neg = (d1 < -eps) or (d2 < -eps) or (d3 < -eps)
        has_pos = (d1 > eps) or (d2 > eps) or (d3 > eps)
        return not (has_neg and has_pos)

    def __draw_segment(self, canvas: Canvas, p1: tuple[float, float], p2: tuple[float, float], **kargs) -> None:
        x1, y1 = int(round(p1[0])), int(round(p1[1]))
        x2, y2 = int(round(p2[0])), int(round(p2[1]))

        if y1 == y2:
            start_x, end_x = min(x1, x2), max(x1, x2)
            canvas.v_line(start_x, y1, end_x - start_x + 1, **kargs)
        elif x1 == x2:
            start_y, end_y = min(y1, y2), max(y1, y2)
            canvas.h_line(x1, start_y, end_y - start_y + 1, **kargs)
        elif y1 < y2:
            canvas.line(x1, y1, x2, y2, **kargs)
            canvas.set_pixel(x2, y2, **kargs)
        else:
            canvas.line(x2, y2, x1, y1, **kargs)
            canvas.set_pixel(x1, y1, **kargs)

    def paint(self, canvas: Canvas, **kargs) -> None:
        v1, v2, v3 = self.__get_vertices()
        self.__draw_segment(canvas, v1, v2, **kargs)
        self.__draw_segment(canvas, v2, v3, **kargs)
        self.__draw_segment(canvas, v3, v1, **kargs)

    def __repr__(self) -> str:
        return (
            f"Triangle(side_a={self.__side_a}, side_b={self.__side_b}, "
            f"side_c={self.__side_c}, x={self.__x}, y={self.__y})"
        )


class CompoundShape(Shape):

    def __init__(self, shapes: list[Shape] | None = None):
        self.__shapes: list[Shape] = list(shapes) if shapes is not None else []

    def add_shape(self, shape: Shape) -> None:
        self.__shapes.append(shape)

    def get_shapes(self) -> list[Shape]:
        return list(self.__shapes)

    def get_x(self) -> float:
        if not self.__shapes:
            return 0.0
        return min(s.get_x() for s in self.__shapes)

    def get_y(self) -> float:
        if not self.__shapes:
            return 0.0
        return min(s.get_y() for s in self.__shapes)

    def area(self) -> float:
        return sum(s.area() for s in self.__shapes)

    def perimeter(self) -> float:
        return sum(s.perimeter() for s in self.__shapes)

    def get_perimeter_points(self) -> list[tuple[float, float]]:
        all_points: list[tuple[float, float]] = []
        for s in self.__shapes:
            all_points.extend(s.get_perimeter_points())
        if len(all_points) <= 16:
            return all_points
        step = len(all_points) / 16.0
        return [all_points[int(i * step)] for i in range(16)]

    def is_inside(self, px: float, py: float) -> bool:
        return any(s.is_inside(px, py) for s in self.__shapes)

    def paint(self, canvas: Canvas, **kargs) -> None:
        """Paints all contained shapes onto the canvas."""
        for shape in self.__shapes:
            shape.paint(canvas, **kargs)

    def __repr__(self) -> str:
        return f"CompoundShape(shapes={self.__shapes})"


# ---  Drawing ---
if __name__ == "__main__":
    canvas = Canvas(width=50, height=22)

    box_outer = Rectangle(length=10, width=18, x=9, y=4)
    box_inner = Rectangle(length=5, width=6, x=14, y=10)
    tri_roof = Triangle(side_a=8, side_b=13, side_c=13, x=1, y=7)
    sun = Circle(radius=5, x=7, y=38)

    scene = CompoundShape([box_outer, box_inner, tri_roof, sun])
    scene.paint(canvas, char="*")

    canvas.display()

    print(f"\nTotal Compound Area:      {scene.area():.2f}")
    print(f"Total Compound Perimeter: {scene.perimeter():.2f}")
    print(f"box_inner overlaps box_outer? {box_inner.overlaps(box_outer)}")
    print(f"sun overlaps box_outer?       {sun.overlaps(box_outer)}")
```

```python

class Canvas:
    def __init__(self, width, height):
        self.width = width
        self.height = height
        self.data = [[' '] * width for i in range(height)]

    def set_pixel(self, row, col, char='*'):
        self.data[row][col] = char

    def get_pixel(self, row, col):
        return self.data[row][col]

    def clear_canvas(self):
        self.data = [[' '] * self.width for i in range(self.height)]

    def v_line(self, x, y, h, **kargs):
        for i in range(x, x + h):
            self.set_pixel(i, y, **kargs)

    def h_line(self, x, y, w, **kargs):
        for i in range(y, y + w):
            self.set_pixel(x, i, **kargs)

    def line(self, x1, y1, x2, y2, **kargs):
        slope = (x2 - x1) / (y2 - y1)
        for y in range(y1, y2):
            x = x1 + int(slope * (y - y1))
            self.set_pixel(x, y, **kargs)

    def display(self):
        print("\n".join(["".join(row) for row in self.data]))


class Shape:

    def __init__(self, **kargs):
        self.__paint_kargs = kargs

    def get_paint_kargs(self) -> dict:
        return dict(self.__paint_kargs)

    def get_x(self) -> float:
        raise NotImplementedError("Subclasses must implement get_x().")

    def get_y(self) -> float:
        raise NotImplementedError("Subclasses must implement get_y().")

    def area(self) -> float:
        raise NotImplementedError("Subclasses must implement area().")

    def perimeter(self) -> float:
        raise NotImplementedError("Subclasses must implement perimeter().")

    def get_perimeter_points(self) -> list[tuple[float, float]]:
        raise NotImplementedError("Subclasses must implement get_perimeter_points().")

    def is_inside(self, px: float, py: float) -> bool:
        raise NotImplementedError("Subclasses must implement is_inside().")

    def overlaps(self, other: "Shape") -> bool:
        for px, py in self.get_perimeter_points():
            if other.is_inside(px, py):
                return True
        for px, py in other.get_perimeter_points():
            if self.is_inside(px, py):
                return True
        return False

    def paint(self, canvas: Canvas, **kargs) -> None:
        raise NotImplementedError("Subclasses must implement paint().")


class Rectangle(Shape):

    def __init__(self, length: float, width: float, x: float = 0.0, y: float = 0.0, **kargs):
        super().__init__(**kargs)
        if length < 0 or width < 0:
            raise ValueError("Length and width must be non-negative.")
        self.__length = length
        self.__width = width
        self.__x = x
        self.__y = y

    def get_length(self) -> float:
        return self.__length

    def get_width(self) -> float:
        return self.__width

    def get_x(self) -> float:
        return self.__x

    def get_y(self) -> float:
        return self.__y

    def area(self) -> float:
        return self.__length * self.__width

    def perimeter(self) -> float:
        return 2 * (self.__length + self.__width)

    def get_perimeter_points(self) -> list[tuple[float, float]]:
        x, y = self.__x, self.__y
        L, W = self.__length, self.__width
        corners = [(x, y), (x + L, y), (x + L, y + W), (x, y + W)]
        points: list[tuple[float, float]] = []
        for i in range(4):
            x1, y1 = corners[i]
            x2, y2 = corners[(i + 1) % 4]
            for step in range(4):
                t = step / 4.0
                points.append((x1 + t * (x2 - x1), y1 + t * (y2 - y1)))
        return points

    def is_inside(self, px: float, py: float) -> bool:
        return (
            self.__x <= px <= self.__x + self.__length
            and self.__y <= py <= self.__y + self.__width
        )

    def paint(self, canvas: Canvas, **kargs) -> None:
        merged_kargs = {**self.get_paint_kargs(), **kargs}
        r = int(round(self.__x))
        c = int(round(self.__y))
        h = int(round(self.__length))
        w = int(round(self.__width))

        canvas.v_line(r, c, h, **merged_kargs)
        canvas.v_line(r, c + w, h, **merged_kargs)
        canvas.h_line(r, c, w, **merged_kargs)
        canvas.h_line(r + h, c, w + 1, **merged_kargs)

    def __repr__(self) -> str:
        return (
            f"Rectangle(length={self.__length}, width={self.__width}, "
            f"x={self.__x}, y={self.__y})"
        )


class Circle(Shape):

    def __init__(self, radius: float, x: float = 0.0, y: float = 0.0, **kargs):
        super().__init__(**kargs)
        if radius < 0:
            raise ValueError("Radius must be non-negative.")
        self.__radius = radius
        self.__x = x
        self.__y = y

    def get_radius(self) -> float:
        return self.__radius

    def get_x(self) -> float:
        return self.__x

    def get_y(self) -> float:
        return self.__y

    def area(self) -> float:
        return math.pi * (self.__radius ** 2)

    def perimeter(self) -> float:
        return 2 * math.pi * self.__radius

    def get_perimeter_points(self) -> list[tuple[float, float]]:
        points: list[tuple[float, float]] = []
        num_points = 16
        for i in range(num_points):
            angle = (2 * math.pi * i) / num_points
            px = self.__x + self.__radius * math.cos(angle)
            py = self.__y + self.__radius * self.__radius * 0 + self.__y + self.__radius * math.sin(angle)
            points.append((px, py))
        return points

    def is_inside(self, px: float, py: float) -> bool:
        return (px - self.__x) ** 2 + (py - self.__y) ** 2 <= self.__radius ** 2

    def paint(self, canvas: Canvas, **kargs) -> None:
        merged_kargs = {**self.get_paint_kargs(), **kargs}
        for px, py in self.get_perimeter_points():
            r = int(round(px))
            c = int(round(py))
            if 0 <= r < canvas.height and 0 <= c < canvas.width:
                canvas.set_pixel(r, c, **merged_kargs)

    def __repr__(self) -> str:
        return f"Circle(radius={self.__radius}, x={self.__x}, y={self.__y})"


class Triangle(Shape):

    def __init__(
        self,
        side_a: float,
        side_b: float,
        side_c: float,
        x: float = 0.0,
        y: float = 0.0,
        **kargs,
    ):
        super().__init__(**kargs)
        if side_a <= 0 or side_b <= 0 or side_c <= 0:
            raise ValueError("All side lengths must be positive.")
        if (
            side_a + side_b <= side_c
            or side_a + side_c <= side_b
            or side_b + side_c <= side_a
        ):
            raise ValueError(
                "Invalid side lengths: violates the triangle inequality theorem."
            )

        self.__side_a = side_a
        self.__side_b = side_b
        self.__side_c = side_c
        self.__x = x
        self.__y = y

    def get_side_a(self) -> float:
        return self.__side_a

    def get_side_b(self) -> float:
        return self.__side_b

    def get_side_c(self) -> float:
        return self.__side_c

    def get_x(self) -> float:
        return self.__x

    def get_y(self) -> float:
        return self.__y

    def __get_vertices(self) -> list[tuple[float, float]]:
        a, b, c = self.__side_a, self.__side_b, self.__side_c
        v1 = (self.__x, self.__y)
        v2 = (self.__x + a, self.__y)

        cos_a = (a**2 + b**2 - c**2) / (2 * a * b)
        cos_a = max(-1.0, min(1.0, cos_a))
        sin_a = math.sqrt(1.0 - cos_a**2)
        v3 = (self.__x + b * cos_a, self.__y + b * sin_a)
        return [v1, v2, v3]

    def area(self) -> float:
        s = self.perimeter() / 2.0
        return math.sqrt(
            s * (s - self.__side_a) * (s - self.__side_b) * (s - self.__side_c)
        )

    def perimeter(self) -> float:
        return self.__side_a + self.__side_b + self.__side_c

    def get_perimeter_points(self) -> list[tuple[float, float]]:
        vertices = self.__get_vertices()
        points: list[tuple[float, float]] = []
        for i in range(3):
            x1, y1 = vertices[i]
            x2, y2 = vertices[(i + 1) % 3]
            for step in range(5):
                t = step / 5.0
                points.append((x1 + t * (x2 - x1), y1 + t * (y2 - y1)))
        return points

    def is_inside(self, px: float, py: float) -> bool:
        (x1, y1), (x2, y2), (x3, y3) = self.__get_vertices()

        def cross_product(ax: float, ay: float, bx: float, by: float) -> float:
            return (bx - ax) * (py - ay) - (by - ay) * (px - ax)

        d1 = cross_product(x1, y1, x2, y2)
        d2 = cross_product(x2, y2, x3, y3)
        d3 = cross_product(x3, y3, x1, y1)

        eps = 1e-9
        has_neg = (d1 < -eps) or (d2 < -eps) or (d3 < -eps)
        has_pos = (d1 > eps) or (d2 > eps) or (d3 > eps)
        return not (has_neg and has_pos)

    def __draw_segment(self, canvas: Canvas, p1: tuple[float, float], p2: tuple[float, float], **kargs) -> None:
        x1, y1 = int(round(p1[0])), int(round(p1[1]))
        x2, y2 = int(round(p2[0])), int(round(p2[1]))

        if y1 == y2:
            start_x, end_x = min(x1, x2), max(x1, x2)
            canvas.v_line(start_x, y1, end_x - start_x + 1, **kargs)
        elif x1 == x2:
            start_y, end_y = min(y1, y2), max(y1, y2)
            canvas.h_line(x1, start_y, end_y - start_y + 1, **kargs)
        elif y1 < y2:
            canvas.line(x1, y1, x2, y2, **kargs)
            canvas.set_pixel(x2, y2, **kargs)
        else:
            canvas.line(x2, y2, x1, y1, **kargs)
            canvas.set_pixel(x1, y1, **kargs)

    def paint(self, canvas: Canvas, **kargs) -> None:
        merged_kargs = {**self.get_paint_kargs(), **kargs}
        v1, v2, v3 = self.__get_vertices()
        self.__draw_segment(canvas, v1, v2, **merged_kargs)
        self.__draw_segment(canvas, v2, v3, **merged_kargs)
        self.__draw_segment(canvas, v3, v1, **merged_kargs)

    def __repr__(self) -> str:
        return (
            f"Triangle(side_a={self.__side_a}, side_b={self.__side_b}, "
            f"side_c={self.__side_c}, x={self.__x}, y={self.__y})"
        )


class CompoundShape(Shape):

    def __init__(self, shapes: list[Shape] | None = None, **kargs):
        super().__init__(**kargs)
        self.__shapes: list[Shape] = list(shapes) if shapes is not None else []

    def add_shape(self, shape: Shape) -> None:
        self.__shapes.append(shape)

    def get_shapes(self) -> list[Shape]:
        return list(self.__shapes)

    def get_x(self) -> float:
        if not self.__shapes:
            return 0.0
        return min(s.get_x() for s in self.__shapes)

    def get_y(self) -> float:
        if not self.__shapes:
            return 0.0
        return min(s.get_y() for s in self.__shapes)

    def area(self) -> float:
        return sum(s.area() for s in self.__shapes)

    def perimeter(self) -> float:
        return sum(s.perimeter() for s in self.__shapes)

    def get_perimeter_points(self) -> list[tuple[float, float]]:
        all_points: list[tuple[float, float]] = []
        for s in self.__shapes:
            all_points.extend(s.get_perimeter_points())
        if len(all_points) <= 16:
            return all_points
        step = len(all_points) / 16.0
        return [all_points[int(i * step)] for i in range(16)]

    def is_inside(self, px: float, py: float) -> bool:
        return any(s.is_inside(px, py) for s in self.__shapes)

    def paint(self, canvas: Canvas, **kargs) -> None:
        merged_kargs = {**self.get_paint_kargs(), **kargs}
        for shape in self.__shapes:
            shape.paint(canvas, **merged_kargs)


class RasterDrawing:

    def __init__(self, shapes: dict[str, Shape] | None = None):
        self.__shapes: dict[str, Shape] = dict(shapes) if shapes is not None else {}
        self.__counter = 0

    def add_shape(self, shape: Shape, name: str | None = None) -> str:
        if name is None:
            name = f"shape_{self.__counter}"
            self.__counter += 1
        self.__shapes[name] = shape
        return name

    def remove_shape(self, name: str) -> None:
        if name in self.__shapes:
            del self.__shapes[name]
        else:
            print(f"Error: Shape '{name}' not found in drawing.")

    def update_shape(self, name: str, new_shape: Shape) -> None:
        self.__shapes[name] = new_shape

    def get_shape(self, name: str) -> Shape | None:
        return self.__shapes.get(name)

    def get_shapes(self) -> dict[str, Shape]:
        return dict(self.__shapes)

    def paint(self, canvas: Canvas, clear: bool = True, **kargs) -> None:
        if clear:
            canvas.clear_canvas()
        for shape in self.__shapes.values():
            shape.paint(canvas, **kargs)

    def paint_and_display(self, canvas: Canvas, clear: bool = True, **kargs) -> None:
        self.paint(canvas, clear=clear, **kargs)
        canvas.display()


if __name__ == "__main__":
    canvas = Canvas(width=50, height=20)
    drawing = RasterDrawing()

    drawing.add_shape(Rectangle(length=9, width=18, x=9, y=4, char="#"), name="walls")
    drawing.add_shape(Rectangle(length=4, width=6, x=14, y=10, char="+"), name="door")
    drawing.add_shape(Triangle(side_a=8, side_b=13, side_c=13, x=1, y=7, char="^"), name="roof")
    drawing.add_shape(Circle(radius=4, x=6, y=38, char="o"), name="sun")

    print("1. Initial Drawing (House with Door, Roof, and Sun)")
    drawing.paint_and_display(canvas)

    drawing.remove_shape("door")
    drawing.update_shape("sun", Circle(radius=5, x=11, y=29, char="@"))

    tree = CompoundShape([
        Rectangle(length=6, width=2, x=12, y=42, char="|"),
        Circle(radius=4, x=8, y=43, char="*")
    ])
    drawing.add_shape(tree, name="tree")

    print("\n2. Modified Drawing (Door Removed, Sun Moved/Resized, Tree Added)")
    drawing.paint_and_display(canvas)
```

   * Overload `__repr__` functions of all objects to return strings of the python code that would construct the object.

For example:

```python
class foo:
    def __init__(self,a,b=None):
        self.a=a
        self.b=b
        
    def __repr__(self):
        return "foo("+repr(self.a)+","+repr(self.b)+")"
    
    def save(self,filename):
        f=open(filename,"w")
        f.write(self.__repr__())
        f.close()
        
   
def foo_loader(filename):
    f=open(filename,"r")
    tmp=eval(f.read())
    f.close()
    return tmp
```
